```python doc = nlp("This is a sentence.")``` creates a  Doc object


The spacy will tokenize the sentence and then call each component on the Doc, in order:
1) tok2vec - this is the token-to-vector component that produces the word vectors which describe the meaning of the words in the context of the sentence.
2) tagger - assigns dependency labels, part-of-speech (POS) tags to words ("run" is a verb, "dog" is a noun, etc.)
3) parser - Builds a syntactic dependency parse tree, identifying subject-verb-object relationships
4) attribute_ruler - Assign token attribute mappings and rule-based exceptions. For example, "U.K." should always be parsed as a single token.
5) lemmatizer - Assign base forms, for example running -> run
6) ner - Detect and label named entities. Named entities are "real world objects" that are assigned a name – for example, a person, an organization or a country.

references <br> https://spacy.io/api/doc <br> https://spacy.io/usage/spacy-101

````python
## Task 2:
## Your code here 
def return_verbs(text:str):

    verb_list = []
    # 1. Load the small English model
    nlp = spacy.load("en_core_web_sm")

    # 2. Process the text
    doc = nlp(text)

    # 3. Print first 5 tokens and part-of-speechs - hint. use token.pos_ tag
    print("First 5 tokens and part-of-speechs:")
    for token in doc[:5]:
        print(token.text,token.pos_)
    
    for token in doc:
        if token.pos_ == 'VERB':
            #print('Verb found')
            verb_list.append(token.text)

        # 4. Find all verbs in doc 
    return verb_list
text = "It’s official: Apple is the first U.S. public company to reach a $1 trillion market value"
verbs = return_verbs(text)
print("\nText verbs:", verbs)
print("Total:", len(verbs))
````

````python
## Task 3:
## Your code here 
def return_lemmas(text:str):
    lemma_list = []
    
    nlp = spacy.load("en_core_web_sm")
    doc = nlp(text)
    print("First 5 tokens and lemmas:")
    for index, token in enumerate(doc):
        if index <5:
            print(token.text,token.lemma_)
        lemma_list.append(token.lemma_)

    return lemma_list
````

````python
text = "I was happy when I ran with my two older dogs."
lemmas = return_lemmas(text)

print("\nText lemmas:", lemmas)
print("Total:", len(lemmas))
````

````python
## Task 4:
## Your code here 
def return_places(text:str):
    place_list = []
    # 1. Load the small English model and process the text
    nlp = spacy.load("en_core_web_sm")
    doc = nlp(text)
    # 2. Print document entity count
    print(f"Document entity count: {len(doc.ents)}")
    # 3. Print document entities
    print("Document entities and entity labels:")
    for ent in doc.ents:
        print(ent.text, ent.label_)
        if ent.label_ == 'GPE':
            place_list.append(ent.text)
    # 4. create list of places (entity label == GPE)
    
    return place_list 
````

````python
text = "It’s official: Apple is the first U.S. public company to reach a $1 trillion market value. Current CEO is Tim Cook."
places = return_places(text)

print("\nText entity places:", places)
print("Total:", len(places))
````

````python
## Task 5:
## Your code here 

def wikipedia_search(text) :

    import spacy
    from spacy.tokens import Span
    
    nlp = spacy.load("en_core_web_sm")

    print("\n")
    
    def get_wikipedia_url(span):
        # Get a Wikipedia URL 
        if span.label_ in ("GPE", "PERSON", "ORG"): ### -1-
            entity_text = span.text.replace(" ", "_")
            return "https://en.wikipedia.org/w/index.php?search=" + entity_text
    
    
    # Set the Span extension wikipedia_url using the getter get_wikipedia_url
    Span.set_extension("wikipedia_url", getter=get_wikipedia_url, force=True)

    urls = []
    doc = nlp(text)
    for ent in doc.ents:
        if ent.label_ in ("GPE", "PERSON", "ORG"): ### -2-
            print('%-15s' % ent.text, '%-10s' % ent.label_, '%-30s' %ent._.wikipedia_url)
            urls.append(ent._.wikipedia_url)
    
    return urls
````

````python
text =  """
        In over fifty years from his very first recordings right through to his 
        last album, David Bowie was at the vanguard of contemporary culture. He lives in England. 
        But does he work for Nokia?
        """

urls=wikipedia_search(text)
urls
````
